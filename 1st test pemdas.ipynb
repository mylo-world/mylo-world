```python
import random

# Function to get a random word from the list
def get_word(word_list):
    return random.choice(word_list).upper()

# Function to display the hangman stages
def display_hangman(tries):
    stages = ["""
                   ------
                   |    |
                   |    O
                   |   \\|/
                   |    |
                   |   / \\
                  ---
                """,
                """
                   ------
                   |    |
                   |    O
                   |   \\|/
                   |    |
                   |   / 
                  ---
                """,
                """
                   ------
                   |    |
                   |    O
                   |   \\|/
                   |    |
                   |    
                  ---
                """,
                """
                   ------
                   |    |
                   |    O
                   |   \\|/
                   |    
                   |    
                  ---
                """,
                """
                   ------
                   |    |
                   |    O
                   |   \\|
                   |    
                   |    
                  ---
                """,
                """
                   ------
                   |    |
                   |    O
                   |    |
                   |    
                   |    
                  ---
                """,
                """
                   ------
                   |    |
                   |    O
                   |    
                   |    
                   |    
                  ---
                """
    ]
    return stages[tries]

# Function to display the vocabulary hint
def get_hint(word, dictionary):
    return dictionary.get(word.lower(), "No hint available.")

def play_hangman(word_list, dictionary):
    word = get_word(word_list)
    guessed = "_" * len(word)
    guessed_correctly = False
    guessed_letters = []
    tries = 6

    print("Let's play Hangman!")
    print(display_hangman(tries))
    print(guessed)
    print("\n")

    while not guessed_correctly and tries > 0:
        guess = input("Please guess a letter: ").upper()
        if len(guess) == 1 and guess.isalpha():
            if guess in guessed_letters:
                print("You already guessed the letter", guess)
            elif guess not in word:
                print(guess, "is not in the word.")
                tries -= 1
                guessed_letters.append(guess)
            else:
                print("Good job,", guess, "is in the word!")
                guessed_letters.append(guess)
                guessed_list = list(guessed)
                indices = [i for i, letter in enumerate(word) if letter == guess]
                for index in indices:
                    guessed_list[index] = guess
                guessed = "".join(guessed_list)
                if "_" not in guessed:
                    guessed_correctly = True
        else:
            print("Not a valid guess.")
        
        print(display_hangman(tries))
        print(guessed)
        print("Hint:", get_hint(word, dictionary))
        print("\n")
    
    if guessed_correctly:
        print("Congratulations, you guessed the word! You win!")
    else:
        print("Sorry, you ran out of tries. The word was " + word + ". Maybe next time!")

if __name__ == "__main__":
    word_list = ["python", "programming", "development", "data", "algorithm"]
    dictionary = {
        "python": "A high-level programming language.",
        "programming": "The process of writing computer programs.",
        "development": "The process of developing software.",
        "data": "Information processed or stored by a computer.",
        "algorithm": "A process or set of rules to be followed in problem-solving operations."
    }
    play_hangman(word_list, dictionary)
```

```python
import tkinter as tk

class App(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("Navigasi Halaman Tkinter")
        
        # Container untuk menyimpan semua frame
        container = tk.Frame(self)
        container.pack(side="top", fill="both", expand=True)
        
        # Dictionary untuk menyimpan referensi ke semua frame
        self.frames = {}
        
        # Inisialisasi semua frame
        for F in (HalamanUtama, HalamanKedua):
            page_name = F.__name__
            frame = F(parent=container, controller=self)
            self.frames[page_name] = frame
            frame.grid(row=0, column=0, sticky="nsew")
        
        self.show_frame("HalamanUtama")
    
    def show_frame(self, page_name):
        '''Menampilkan frame yang diinginkan'''
        frame = self.frames[page_name]
        frame.tkraise()

class HalamanUtama(tk.Frame):
    def __init__(self, parent, controller):
        super().__init__(parent)
        self.controller = controller
        
        label = tk.Label(self, text="Ini adalah Halaman Utama", font=("Helvetica", 16))
        label.pack(pady=10, padx=10)
        
        button = tk.Button(self, text="Ke Halaman Kedua",
                           command=lambda: controller.show_frame("HalamanKedua"))
        button.pack()

class HalamanKedua(tk.Frame):
    def __init__(self, parent, controller):
        super().__init__(parent)
        self.controller = controller
        
        label = tk.Label(self, text="Ini adalah Halaman Kedua", font=("Helvetica", 16))
        label.pack(pady=10, padx=10)
        
        button = tk.Button(self, text="Kembali ke Halaman Utama",
                           command=lambda: controller.show_frame("HalamanUtama"))
        button.pack()

if __name__ == "__main__":
    app = App()
    app.mainloop()
```

```python
import tkinter as tk
from PIL import Image, ImageTk, ImageSequence

class AnimatedGIF(tk.Label):
    def __init__(self, master, gif_path, *args, **kwargs):
        super().__init__(master, *args, **kwargs)
        self.gif_path = gif_path
        self.frames = [ImageTk.PhotoImage(img) for img in ImageSequence.Iterator(Image.open(gif_path))]
        self.frame_index = 0
        self.total_frames = len(self.frames)
        self.update_gif()
    
    def update_gif(self):
        self.config(image=self.frames[self.frame_index])
        self.frame_index = (self.frame_index + 1) % self.total_frames
        self.after(100, self.update_gif)  # Update frame every 100 milliseconds

root = tk.Tk()
root.title("Gambar Bergerak di Tkinter")

# Tambahkan widget animasi GIF
gif_label = AnimatedGIF(root, r"C:\Users\LOQ\Python latihan\college studies\assignment\PROJECT\Assets\OciHCvJ.gif")
gif_label.pack()

root.mainloop()
```
